# List And Loops 

```python
marks = [65, 88, 92, 74, 81]

highest = marks[0]
for m in marks:
    if m > highest:
        highest = m

print("The highest marks are:", highest)
```

```python
num = int(input("Enter a number to print its multiplication table: "))

for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

# Function

```python
def check_number(num):
    if num > 0:
        return "Positive Number"
    elif num < 0:
        return "Negative Number"
    else:
        return "Zero"

result = check_number(1)
print("Result:", result)
```

```python
def even_or_odd(number):
    if number % 2 == 0:
        return "Even Number"
    else:
        return "Odd Number"

print(even_or_odd(14))
print(even_or_odd(9))
```

```python
def get_grade(marks):
    if marks >= 80:
        return "A+ Grade"
    elif marks >= 70:
        return "A Grade"
    elif marks >= 60:
        return "B Grade"
    else:
        return "Fail - Try again"

student_marks = int(input("Enter your marks: "))
print("Your Result:", get_grade(student_marks))
```
